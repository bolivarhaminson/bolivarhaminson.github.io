```python
#Pide una frase o palabra y muestra su version invertida

texto = input("ingresa una palabra o una frase:")
invertida = ""
for caracter in texto:
    invertida = caracter + invertida

print(invertida)
```

```python
# suma de matrizes 2x2

A=[[int(input(f"A{i}{j}:")) for j in range(2)] for i in range(2)]
B=[[int(input(f"B{i}{j}:")) for j in range(2)] for i in range(2)]
print("Resultado:",[[A[i][j]+ B[i][j] for j in range(2)] for i in range(2)])

# integrantes: Christian Barros y Haminson Bolivar
```
